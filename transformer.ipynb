```python
from torch import Tensor
# 将输入的词汇表转化为指定维度的Embedding

class TokenEmbedding(nn.Embedding):
    def __init__(self, vocab_size, d_model):
        super(TokenEmbedding, self).__init__(vocab_size, d_model)
```

```python
class PositionEmbedding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionEmbedding, self).__init__()
        self.encoding=torch.zeros(max_len,d_model)
        self.encoding.requires_grad=False
        pos=torch.arange(0,max_len)
        pos=pos.float().unsqueeze(dim=1)
        _2i=torch.arange(0,d_model,step=2).float()
        self.encoding[:,0::2]=torch.sin(pos/10000**(_2i/d_model))
        self.encoding[:,1::2]=torch.cos(pos/10000**(_2i/d_model))

    def forward(self,x:Tensor)->Tensor:
        batch_size, seq_len = x.size()
        return self.encoding[:seq_len, :]
```

```python
class TransformerEmbedding(nn.Module):
    def __init__(self, vocab_size, d_model, max_len=5000, dropout=0.1):
        super(TransformerEmbedding, self).__init__()
        self.token_embedding=TokenEmbedding(vocab_size,d_model)
        self.position_embedding=PositionEmbedding(d_model,max_len)
        self.dropout=nn.Dropout(dropout)

    def forward(self,x:Tensor)->Tensor:
        return self.token_embedding(x)+self.position_embedding(x)
```
